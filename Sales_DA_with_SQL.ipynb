```python
# Applying Essential SQL Queries for Sales Analysis on the Global Superstore Dataset

# In this work, I implemented a set of key SQL queries to explore and analyze sales data from the Global Superstore dataset.
# These queries cover important aspects such as country-wise sales, product categories, profit analysis, shipping trends, and customer segmentation.
# The results provide valuable insights into sales performance and help identify opportunities for business optimization.

# Dataset source: https://github.com/yannie28/Global-Superstore/blob/master/Global_Superstore_100.csv
```

```python
import sqlite3
import pandas as pd

# Create a new SQLite database connection (or open it if it already exists)
conn = sqlite3.connect('my_database.db')

# Create a cursor object to execute SQL commands
cursor = conn.cursor()
```

```python

# Load the cleaned housing data from a CSV file into a pandas DataFrame
dff = pd.read_csv('/content/Global_Superstore_100.csv')

pd.set_option('display.max_columns', None)

# Display the first few rows of the DataFrame to verify successful loading
dff.head()
```

```python
# Transfer data from the pandas DataFrame to the SQLite database table named 'orders'.
# If the table already exists, replace it with the new data.
# Do not include the DataFrame's index as a column in the table.
dff.to_sql('orders', conn, if_exists='replace', index=False)

# Execute an SQL query to select the first 10 rows from the 'orders' table
cursor.execute("SELECT * FROM orders LIMIT 10")

# Fetch all rows returned by the query
rows = cursor.fetchall()

# Loop through the fetched rows and print each one
for row in rows:
    print(row)
```

```python
# Retrieve metadata about the 'orders' table structure (such as column names, data types, etc.)
cursor.execute("PRAGMA table_info(orders);")

# Fetch all metadata rows returned by the PRAGMA command
columns = cursor.fetchall()

# Loop through the metadata and print each column's details
for col in columns:
    print(col)

# Based on the output, the column names and structure are correctly set in the SQLite table,
# confirming that the data was imported successfully.
```

```python
# This approach combines the readability of Python with the power of SQL.
def run_sqls(*queries):
    results = []
    for q in queries:
        df = pd.read_sql_query(q, conn)  # Run each SQL query and return result as a DataFrame
        results.append(df)               # Store the result in the results list
    return results                       # Return all query results as a list of DataFrames
```

```python
# Example usage of the run_sqls function:


# Define multiple SQL queries to be executed on the 'orders' table:

# Query 1: Select all orders where the country is 'United States'
q1 = "SELECT * FROM orders WHERE Country = 'United States'"

# Query 2: Select all products where the category is 'Furniture'
q2 = "SELECT * FROM orders WHERE Category = 'Furniture'"

# Query 3: Calculate total sales grouped by country, sorted in descending order of sales
q3 = "SELECT Country, SUM(Sales) AS Total_Sales FROM orders GROUP BY Country ORDER BY Total_Sales DESC"

# Query 4: Get cities with negative total profit (loss), ordered by total profit ascending
q4 = "SELECT City, SUM(Profit) AS Total_Profit FROM orders GROUP BY City HAVING SUM(Profit) < 0 ORDER BY Total_Profit"

# Query 5: Calculate the average discount for each product category
q5 = "SELECT Category, AVG(Discount) AS Avg_Discount FROM orders GROUP BY Category"

# Query 6: Count the number of orders per year by extracting the year from 'Order Date'
q6 = """
SELECT strftime('%Y', "Order Date") AS Year, COUNT(*) AS Num_Orders
FROM orders
GROUP BY Year
ORDER BY Year
"""

# Query 7: Find the best-selling product by total sales
q7 = """
SELECT "Product Name", SUM(Sales) AS Total_Sales
FROM orders
GROUP BY "Product Name"
ORDER BY Total_Sales DESC
LIMIT 3
"""

# Query 8: Find the most profitable customer segment by total profit
q8 = """
SELECT Segment, SUM(Profit) AS Total_Profit
FROM orders
GROUP BY Segment
ORDER BY Total_Profit DESC
LIMIT 3
"""

# Query 9: Find the month with the highest number of shipments by extracting month from 'Ship Date'
q9 = """
SELECT strftime('%m', "Ship Date") AS Ship_Month, COUNT(*) AS Num_Shipments
FROM orders
GROUP BY Ship_Month
ORDER BY Num_Shipments DESC
LIMIT 2
"""

# Query 10: Calculate the average shipping cost per market, ordered descending by average shipping cost
q10 = """
SELECT Market, AVG("Shipping Cost") AS Avg_Shipping_Cost
FROM orders
GROUP BY Market
ORDER BY Avg_Shipping_Cost DESC
"""

# Pass all these queries to the function run_sqls, which executes them all
# and returns a list of pandas DataFrames—one DataFrame per query.
results = run_sqls(q1, q2, q3, q4, q5, q6, q7, q8, q9, q10)

# The results variable now holds the outputs from all queries,
# which can be iterated over or accessed individually for further analysis.
```

```python
# Titles corresponding to each query above, used for labeling the output
query_titles = [
  "Orders from the United States",
    "Products in the Furniture category",
    "Total sales by country",
    "Cities with negative (loss) profits",
    "Average discount per category",
    "Number of orders per year",
    "Best-selling product",
    "Most profitable segment",
    "Month with the highest number of shipments",
    "Average shipping cost per market"
]

# Display results with titles
for i, df in enumerate(results):
    print(f"\nResult {i+1}: {query_titles[i]}")  # Print result number and Arabic title
    display(df)  # Display DataFrame (works in Jupyter)
```

```python
# Close the database connection when all queries and operations are done
conn.close()
```
